# Chapter 6: Functions and Reusability

Let's learn how to package up code into functions, so we can reuse it instead of repeating ourselves.

## 1. Defining a Function

A function is a named block of code you can run whenever you need it. You create one with the word `def`, followed by a name and parentheses.

**Task:** Define a function called `greet` that prints `"Hello!"`, then call it.

```python
# Your code here
def greet():
    print("hello world")
greet()
```

## 2. Calling a Function Many Times

Once a function is defined, you can call it as many times as you like.

**Task:** Call the `greet` function three times.

```python
# Your code here
greet()
greet()
greet()
```

## 3. Passing Information with a Parameter

A parameter is a name inside the parentheses that lets you send information into a function.

**Task:** Define a function called `greet_person` that takes a `name` and prints a greeting using it.

```python
# Your code here
def greet_person(name):
    print("name",name)
greet_person("Sam")
```

## 4. Using Several Parameters

A function can take more than one piece of information, separated by commas.

**Task:** Define a function called `greet_with_age` that takes a `name` and an `age`, then prints a sentence using both.

```python
# Your code here
def greet_with_age(name,age):
    print("name",name,"age",age)
    print(f"name = {name} age = {age}")
greet_with_age("Sam",20)
```

 ## 5. Sending Back and Reusing a Result with return

`return` lets a function send a value back to whoever called it, and that value can be stored in a variable and reused later, just like any other value.

**Task:** Define a function called `add_numbers` that takes two numbers and returns their sum. Store the result in a variable and print it, then use that variable in another calculation and print the new result.

```python
# Your code here
def add_numbers(a,b):
    c = a + b
    return c
sum_of_values = add_numbers(10,20)
print(sum_of_values)
```

## 6. Giving a Parameter a Default Value

A default value is used automatically if the caller doesn't provide one for that parameter.

**Task:** Define a function called `greet_default` that takes a `name` with a default value of `"friend"`, then call it without an argument.

```python
# Your code here
def greet_default(name="friend"):
    print("hello",name)
greet_default()
```

## 7. Overriding a Default Value

You can still pass your own value for a parameter that has a default - your value replaces it.

**Task:** Call `greet_default` again, this time passing a name of your own.

```python
# Your code here
greet_default("Sam")
```

## 8. Naming Arguments with Keywords

You can name which parameter each value belongs to when calling a function - this is called a keyword argument, and it lets you list them in any order.

**Task:** Call `greet_with_age` using keyword arguments, in a different order than the function defines them.

```python
# Your code here
def greet_with_age(name,age):
    print(f"name = {name} age = {age}")
greet_with_age(age=21,name="Sam")
```

## 9. Accepting Any Number of Values with *args

Writing `*args` before a parameter name lets a function accept any number of values, gathered together into a tuple.

**Task:** Define a function called `add_all` that takes `*numbers` and returns their total using `sum()`.

```python
# Your code here
def add_all(*numbers):
    return sum(numbers)
print(add_all(10,20,30,40,50))
```

## 10. Accepting Any Number of Named Values with **kwargs

Writing `**kwargs` lets a function accept any number of keyword arguments, gathered together into a dictionary.

**Task:** Define a function called `show_details` that takes `**details` and prints each key and value.

```python
# Your code here
def show_details(**details):
    for key,value in details.items():
        print(key,value)
show_details(name="Sam",age=20)
```

## 11. Variables Inside a Function Stay Inside

A variable created inside a function only exists while that function is running, and can't be used outside it.

**Task:** Define a function that creates a variable called `message` and prints it, then call the function.

```python
# Your code here
def func():
    message = "hello world"
    print(message)
func()

```

## 12. Reading a Variable from Outside the Function

A function can read a variable that was created outside it, as long as it doesn't try to change it.

**Task:** Create a variable called `city` outside any function, then define a function that prints it.

```python
# Your code here
city = "Pune"
def get_city():
    return city
get_city()
```

## 13. Changing an Outside Variable with global

To change a variable from outside the function, instead of creating a new local one, use the word `global` inside the function.

**Task:** Create a variable called `counter` set to `0`, then define a function that uses `global` to increase it by `1`. Call the function and print `counter`.

```python
# Your code here
counter = 0
def increment():
    global counter
    counter += 1
    print(counter)
increment()
```

## 14. Writing a Short Function with lambda

A lambda is a small, unnamed function written on a single line, useful for short pieces of logic.

**Task:** Store a lambda in a variable called `square` that returns a number multiplied by itself, then call it.

```python
# Your code here
square  = lambda x:x*x
print(square(9))

def add(x):
    return x+x
print(add(5))
```

## 15. A Lambda with Two Inputs

A lambda can take more than one input, separated by commas, just like a regular function.

**Task:** Store a lambda called `multiply` that takes two numbers and returns their product, then call it.

```python
# Your code here
multiply = lambda x,y:x*y
print(multiply(3,4))
```

## 16. Applying a Function to Every Item with map()

`map()` runs a function on every item in a list and gives back the results.

**Task:** Use `map()` with the `square` lambda to square every number in a list, then turn the result into a list and print it.

```python
# Your code here
numbers = [1,2,3,4,5]
square_values = list(map(square,numbers))
print(square_values)
```

## 17. Using map() with a Lambda Directly

Instead of storing a lambda first, you can write it directly inside `map()`.

**Task:** Use `map()` with a lambda directly to add `10` to every number in a list.

```python
# Your code here
numbers = [1,2,3,4]
squared = list(map(lambda x: x*x,numbers))
print(squared)
```

## 18. Keeping Only Matching Items with filter()

`filter()` keeps only the items from a list for which a function returns `True`.

**Task:** Use `filter()` with a lambda to keep only the even numbers from a list.

```python
# Your code here
numbers = [1,2,3,4]
filtered = list(filter(lambda x: x%2==0,numbers))
print(filtered)
```

## 19. Combining All Items into One Value with reduce()

`reduce()` combines every item in a list into a single value, applying a function step by step. It comes from the `functools` module, so it needs to be imported first.

**Task:** Import `reduce` from `functools`, then use it with a lambda to multiply every number in a list together.

```python
# Your code here
from functools import reduce
numbers = [1,2,3,4,5]
reduced = reduce(lambda x,y:x*y,numbers)
print(reduced)
```

## 20. Putting It Together: Reusable Functions for a List of Prices

Let's combine functions, default values, and `map()`/`filter()`/`reduce()` in one practical example.

**Task:**
1. Define a function called `apply_discount` that takes a `price` and a `discount` with a default value of `10`, and returns the price after subtracting that percentage.
2. Store a list of prices, then use `map()` with `apply_discount` to get the discounted prices, using the default discount.
3. Use `filter()` to keep only the discounted prices above `50`.
4. Use `reduce()` to add up the discounted prices that were kept, and print the total.

```python
# Your code here
def apply_discount(price,discount=10):
    return price - (price*discount/100)

list_of_prices = [40,50,60,70]
discounted_price = list(map(apply_discount,list_of_prices))
print(discounted_price)

high_prices = list(filter(lambda x: x>50,discounted_price))
print(high_prices)

total = reduce(lambda x,y: x+y,high_prices)
print(total)
```

## Recap

- Functions are defined with `def`, can take parameters, and use `return` to send back a result.
- Parameters can have default values, and arguments can be passed by position or by keyword.
- `*args` collects extra positional values into a tuple, and `**kwargs` collects extra keyword values into a dictionary.
- Variables created inside a function are local to it, but `global` lets a function change a variable from outside.
- `lambda` creates small, unnamed functions, often used with `map()`, `filter()`, and `reduce()` to transform, select, and combine items in a list.
