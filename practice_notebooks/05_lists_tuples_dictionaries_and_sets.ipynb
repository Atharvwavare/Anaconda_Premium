## 31. Putting It Together: Building Records from Lists

Let's combine lists, dictionaries, sets, and pairing tools in one practical example.

**Task:**
1. Store a list of names and a matching list of countries, with some countries repeated.
2. Use `zip()` to build a list of dictionaries, one per person, each with a `"name"` and a `"country"` key.
3. Pull the country out of every record into a list, then use a set to find the unique countries, sorted.
4. Use `enumerate()` to print each person's record with a number in front of it.

```python
# Your code here
names = ["Sam","Peter","Smith","Alen"]
countries = ["India","USA","Germany","India"]

records = []
for name,country in zip(names,countries):
    print(name,country)
    new_dict = {"name":name,"country":country}
    print(new_dict)
    records.append(new_dict)

print(records)


countries_new  = [record["country"] for record in records]
print(countries_new)

unique_countries = set(countries)
print(unique_countries)

for index,value in enumerate(records):
    print(index,value)


```

## Recap

- Lists store ordered, changeable collections of items - indexing, slicing, and methods like `append()`, `pop()`, and `sort()` all work with them.
- List comprehensions build lists in one line, and `enumerate()`/`zip()` help loop through one or more lists together.
- Tuples are like lists but immutable - their items can't be changed after they're created.
- Dictionaries store key-value pairs, with `get()`, `keys()`, `values()`, and `items()` for working with them, and can be nested inside each other.
- Sets store unique, unordered items, and support union (`|`), intersection (`&`), and difference (`-`).
